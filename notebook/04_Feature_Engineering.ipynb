```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\aiml-project\data\processed\cleaned_aqi.csv")
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
df["Date"] = pd.to_datetime(df["Date"])

print("Minimum Date:", df["Date"].min())
print("Maximum Date:", df["Date"].max())
```

```python
df = df.sort_values(
    by=["City", "Date"]
).reset_index(drop=True)
df.head(20)
```

```python
df["Year"] = df["Date"].dt.year
df[["Date", "Year"]].head()
```

```python
df["Month_Number"] = df["Date"].dt.month
df[["Date", "Month_Number"]].head()
```

```python
df["Month_Name"] = df["Date"].dt.month_name()
df[["Date", "Month_Number", "Month_Name"]].head(12)
```

```python
df["Day"] = df["Date"].dt.day
df[["Date", "Day"]].head()
```

```python
df["Day_of_Week"] = df["Date"].dt.dayofweek
df[["Date", "Day_of_Week"]].head(10)
```

```python
df["Quarter"] = df["Date"].dt.quarter
df[["Date", "Quarter"]].head(10)
```

```python
time_features = [
    "Date",
    "Year",
    "Month_Number",
    "Month_Name",
    "Day",
    "Day_of_Week",
    "Quarter"
]

df[time_features].head(20)
```

```python
print("Years:")
print(sorted(df["Year"].unique()))

print("\nMonth Numbers:")
print(sorted(df["Month_Number"].unique()))

print("\nMonth Names:")
print(df["Month_Name"].unique())

print("\nDay of Week:")
print(sorted(df["Day_of_Week"].unique()))

print("\nQuarters:")
print(sorted(df["Quarter"].unique()))
```

```python
df["AQI_Lag_1"] = (
    df.groupby("City")["AQI"]
    .shift(1)
)

df[
    [
        "City",
        "Date",
        "AQI",
        "AQI_Lag_1"
    ]
].head(20)
```

```python
df["AQI_Lag_7"] = (
    df.groupby("City")["AQI"]
    .shift(7)
)

df[
    [
        "City",
        "Date",
        "AQI",
        "AQI_Lag_1",
        "AQI_Lag_7"
    ]
].head(20)
```

```python
df["AQI_Rolling_3"] = (
    df.groupby("City")["AQI"]
    .transform(
        lambda x: x.shift(1).rolling(window=3).mean()
    )
)
```

```python
df["AQI_Rolling_7"] = (
    df.groupby("City")["AQI"]
    .transform(
        lambda x: x.shift(1).rolling(window=7).mean()
    )
)
```

```python
historical_features = [
    "City",
    "Date",
    "AQI",
    "AQI_Lag_1",
    "AQI_Lag_7",
    "AQI_Rolling_3",
    "AQI_Rolling_7"
]

df[historical_features].head(20)
```

```python
city_name = "Delhi"
df[
    df["City"] == city_name
][
    [
        "City",
        "Date",
        "AQI",
        "AQI_Lag_1",
        "AQI_Lag_7",
        "AQI_Rolling_3",
        "AQI_Rolling_7"
    ]
].head(20)
```

```python
city_name = "Delhi"

city_data = (
    df[df["City"] == city_name]
    .sort_values("Date")
    .copy()
)

city_data[
    [
        "Date",
        "AQI",
        "AQI_Lag_1",
        "AQI_Rolling_3",
        "AQI_Rolling_7"
    ]
].head(15)
```

```python
feature_columns = [
    "AQI_Lag_1",
    "AQI_Lag_7",
    "AQI_Rolling_3",
    "AQI_Rolling_7"
]

df[feature_columns].isnull().sum()
```

```python
df = df.dropna(
    subset=["AQI_Lag_7"]
).reset_index(drop=True)

print("Shape after removing insufficient historical records:")
print(df.shape)
```
